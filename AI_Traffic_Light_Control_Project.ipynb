# AI/ML-Based Intelligent Traffic-Light Control

## Project Problem and Intended AI/ML Outcome

Urban traffic congestion causes increased vehicle waiting times, longer queues,
higher fuel consumption, increased travel time, and environmental pollution.
Many traffic-light systems use fixed timing schedules and may not respond
effectively to changing traffic conditions.

This project investigates how Artificial Intelligence (AI) and Machine Learning
(ML) can be used to predict traffic conditions and support intelligent
traffic-light control.

The intended AI/ML outcome is to develop a model that can use historical and
real-time traffic information to predict traffic flow or congestion. These
predictions could then be used by an intelligent traffic-control system to
dynamically adjust traffic-light timing in order to reduce waiting time,
queue lengths, unnecessary stops, and overall travel time.

For this assignment, the machine-learning component will focus on predicting
traffic volume/flow from available traffic and time-related features.


## Data Requirements

## Dataset

The dataset selected for this project is the **Metro Interstate Traffic Volume**
dataset from the UCI Machine Learning Repository.

The dataset contains hourly traffic observations from Interstate 94 in the
Minneapolis-St. Paul area from 2012 to 2018. It contains traffic volume,
weather conditions, holidays, and date/time information.

The dataset contains 48,204 observations and 9 variables. The main target
variable is `traffic_volume`, which represents the hourly number of vehicles
recorded at the monitoring station.

### Dataset Source

UCI Machine Learning Repository:
Metro Interstate Traffic Volume

The dataset is suitable for this project because traffic volume can be used as
an indicator of traffic demand. Predicting traffic volume can help an
intelligent traffic-control system anticipate changing traffic conditions and
make better traffic-light timing decisions.
