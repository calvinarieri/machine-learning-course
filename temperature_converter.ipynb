# Temperature Conversion Assignment

This program converts temperatures from Celsius (°C) to Fahrenheit (°F), checks temperatures against a threshold, and demonstrates the use of a loop and a function.


```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

celsius = float(input("Enter a temperature in °C: "))

fahrenheit = celsius_to_fahrenheit(celsius)

print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")

threshold = 30

if celsius > threshold:
    print(f"The temperature is above the {threshold}°C threshold.")
elif celsius < threshold:
    print(f"The temperature is below the {threshold}°C threshold.")
else:
    print(f"The temperature is exactly at the {threshold}°C threshold.")

temperatures = [0, 10, 20, 30, 40]

print("\nTemperature conversions:")

for temp in temperatures:
    converted = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {converted:.2f}°F")
```
